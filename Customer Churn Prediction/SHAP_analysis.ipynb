```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import shap
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
import warnings
warnings.filterwarnings("ignore")
 
plt.rcParams.update({"figure.facecolor": "white", "axes.spines.top": False,
                     "axes.spines.right": False, "axes.grid": True,
                     "grid.color": "#e5e5e5"})
 
SEED = 42
shap.initjs() 
```

### LOAD DATA & RETRAIN XGBoost ON PREPROCESSED FEATURES

```python
#    SHAP TreeExplainer works directly on the XGBoost model.
#    We retrain here on the preprocessed matrices saved from Phase 2.
# ════════════════════════════════════════════════════════════════════════════
 
print("=" * 60)
print("PHASE 3 — SHAP EXPLAINABILITY")
print("=" * 60)
 
# Load preprocessed test set (features + predictions from Phase 2)
test_df  = pd.read_csv("churn_test_predictions.csv")
train_df = pd.read_csv("churn_train_preprocessed.csv")
 
# Separate target and prediction columns
y_test   = test_df["y_true"].values
y_prob   = test_df["y_prob"].values
y_pred   = test_df["y_pred"].values
 
# Feature matrix — drop non-feature columns
PRED_COLS    = ["y_true", "y_prob", "y_pred"]
X_test_proc  = test_df.drop(columns=PRED_COLS).values
X_train_proc = train_df.values
feature_names = test_df.drop(columns=PRED_COLS).columns.tolist()
 
print(f"Test samples      : {len(X_test_proc)}")
print(f"Train samples     : {len(X_train_proc)}")
print(f"Features          : {len(feature_names)}")
print(f"Churn rate (test) : {y_test.mean()*100:.1f}%")
 
# Retrain XGBoost on preprocessed training data
# Load original target for training
df_raw = pd.read_csv("telco_clean.csv")
df_raw = df_raw.drop(columns=["customerID","TenureBand"], errors="ignore")
y_full = (df_raw["Churn"] == "Yes").astype(int)
 
_, _, y_train, _ = train_test_split(
    df_raw, y_full, test_size=0.2, stratify=y_full, random_state=SEED
)
 
xgb_model = XGBClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=5,
    subsample=0.8, colsample_bytree=0.8,
    scale_pos_weight=(y_train==0).sum()/(y_train==1).sum(),
    random_state=SEED, eval_metric="logloss", verbosity=0,
)
xgb_model.fit(X_train_proc, y_train)
print(f"\nXGBoost retrained on preprocessed features.")
```
